## Dataset
Place `sales_data_sample.csv` in `../datasets/`. The notebook selects numeric columns automatically. You can set `FEATURE_COLUMNS` to the numeric sales features required by your lab.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score

DATA_PATH = "../datasets/sales_data_sample.csv"
df = pd.read_csv(DATA_PATH, encoding="latin1")
print(df.head())
numeric = df.select_dtypes(include="number")
FEATURE_COLUMNS = numeric.columns.tolist()  # edit to select specific columns
X = numeric[FEATURE_COLUMNS].dropna()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("Features used:", FEATURE_COLUMNS)
```

```python
inertias = []
k_values = range(1, 11)
for k in k_values:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled)
    inertias.append(km.inertia_)

plt.plot(list(k_values), inertias, marker="o")
plt.xlabel("Number of clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.grid(True)
plt.show()
```

```python
# Choose k after inspecting the elbow plot.
k = 3
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_scaled)
print("K-Means cluster counts:")
print(pd.Series(kmeans_labels).value_counts().sort_index())
if len(set(kmeans_labels)) > 1:
    print("Silhouette score:", silhouette_score(X_scaled, kmeans_labels))

hierarchical = AgglomerativeClustering(n_clusters=k, linkage="ward")
hier_labels = hierarchical.fit_predict(X_scaled)
print("\nHierarchical cluster counts:")
print(pd.Series(hier_labels).value_counts().sort_index())
```

## Notes
The elbow plot helps select k; k=3 is only an example and should be changed based on the plot and your lab instructions. Standardization is applied before distance-based clustering.
