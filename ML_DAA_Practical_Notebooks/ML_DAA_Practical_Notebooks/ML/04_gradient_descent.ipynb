## Aim
Minimize f(x) = (x + 3)² using gradient descent, starting at x = 2.

```python
import matplotlib.pyplot as plt

def f(x):
    return (x + 3) ** 2

def gradient(x):
    return 2 * (x + 3)

learning_rate = 0.1
x = 2.0
iterations = 30
history = []

for step in range(iterations):
    history.append((step, x, f(x)))
    x = x - learning_rate * gradient(x)

print("Approximate minimum x:", x)
print("Function value:", f(x))
```

```python
xs = [item[1] for item in history]
plt.plot(range(len(xs)), xs, marker="o")
plt.xlabel("Iteration")
plt.ylabel("x")
plt.title("Gradient Descent Convergence")
plt.grid(True)
plt.show()

grid = [i / 10 for i in range(-60, 31)]
plt.plot(grid, [f(v) for v in grid])
plt.scatter(xs, [f(v) for v in xs])
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("f(x) = (x + 3)^2")
plt.grid(True)
plt.show()
```

## Result
The minimum occurs at x = -3 and f(x) = 0. Gradient descent approaches this point when the learning rate is suitably chosen.
