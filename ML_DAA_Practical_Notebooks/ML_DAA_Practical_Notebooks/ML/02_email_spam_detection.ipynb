## Dataset
Place `email.csv` in `../datasets/`. Expected columns: `email` (message text) and `label` (spam/ham or 1/0). Change the names below if needed.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.svm import LinearSVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

DATA_PATH = "../datasets/email.csv"
TEXT_COLUMN = "email"
LABEL_COLUMN = "label"
df = pd.read_csv(DATA_PATH).dropna(subset=[TEXT_COLUMN, LABEL_COLUMN])
X_train, X_test, y_train, y_test = train_test_split(
    df[TEXT_COLUMN].astype(str), df[LABEL_COLUMN],
    test_size=0.2, random_state=42, stratify=df[LABEL_COLUMN]
)
print(df[LABEL_COLUMN].value_counts())
```

```python
models = {
    "Linear SVM": Pipeline([
        ("tfidf", TfidfVectorizer(stop_words="english", max_features=10000)),
        ("classifier", LinearSVC())
    ]),
    "KNN": Pipeline([
        ("tfidf", TfidfVectorizer(stop_words="english", max_features=3000)),
        ("classifier", KNeighborsClassifier(n_neighbors=5))
    ])
}
for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    print("\n", name)
    print("Accuracy:", accuracy_score(y_test, pred))
    print("Confusion matrix:\n", confusion_matrix(y_test, pred))
    print(classification_report(y_test, pred))
```

## Notes
TF-IDF converts text into numeric features. SVM and KNN are trained and evaluated on the same held-out test split.
