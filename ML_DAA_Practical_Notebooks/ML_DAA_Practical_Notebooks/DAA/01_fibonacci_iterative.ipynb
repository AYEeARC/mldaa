## Aim
Generate Fibonacci numbers using an iterative method and count loop iterations.

```python
def fibonacci_iterative(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    a, b = 0, 1
    sequence = []
    steps = 0
    for _ in range(n):
        sequence.append(a)
        a, b = b, a + b
        steps += 1
    return sequence, steps

n = 10
values, steps = fibonacci_iterative(n)
print("Fibonacci sequence:", values)
print("Loop iterations:", steps)
```

## Complexity
- Time: O(n)
- Auxiliary space: O(1), excluding the output list; O(n) including the returned sequence.
