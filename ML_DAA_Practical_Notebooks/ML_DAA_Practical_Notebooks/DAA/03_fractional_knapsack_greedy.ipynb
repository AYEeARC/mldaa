## Aim
Maximize profit in a knapsack when fractions of items are allowed.

```python
def fractional_knapsack(items, capacity):
    # Each item is (weight, profit)
    if capacity < 0:
        raise ValueError("Capacity must be non-negative")
    items = sorted(items, key=lambda item: item[1] / item[0], reverse=True)
    total_profit = 0.0
    chosen = []

    for weight, profit in items:
        if capacity == 0:
            break
        if weight <= 0:
            raise ValueError("Item weights must be positive")
        fraction = min(1.0, capacity / weight)
        total_profit += profit * fraction
        chosen.append((weight, profit, fraction))
        capacity -= weight * fraction

    return total_profit, chosen

items = [(10, 60), (20, 100), (30, 120)]
capacity = 50
profit, selected = fractional_knapsack(items, capacity)
print("Selected (weight, profit, fraction):", selected)
print("Maximum profit:", profit)
```

## Complexity
Sorting: O(n log n). Selection: O(n). Overall: O(n log n).
