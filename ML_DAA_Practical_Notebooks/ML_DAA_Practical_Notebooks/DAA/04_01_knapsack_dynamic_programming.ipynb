## Aim
Solve the 0/1 Knapsack problem using dynamic programming. Each item is either selected completely or not selected.

```python
def knapsack_01(weights, profits, capacity):
    if len(weights) != len(profits):
        raise ValueError("weights and profits must have equal length")
    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        weight = weights[i - 1]
        profit = profits[i - 1]
        for cap in range(capacity + 1):
            dp[i][cap] = dp[i - 1][cap]
            if weight <= cap:
                dp[i][cap] = max(dp[i][cap],
                                 profit + dp[i - 1][cap - weight])

    selected = []
    cap = capacity
    for i in range(n, 0, -1):
        if dp[i][cap] != dp[i - 1][cap]:
            selected.append(i - 1)
            cap -= weights[i - 1]

    selected.reverse()
    return dp[n][capacity], selected, dp

weights = [10, 20, 30]
profits = [60, 100, 120]
capacity = 50
maximum_profit, selected_indices, table = knapsack_01(weights, profits, capacity)
print("Maximum profit:", maximum_profit)
print("Selected item indices (0-based):", selected_indices)
print("Selected weights:", [weights[i] for i in selected_indices])
```

## Complexity
- Time: O(nW), where W is capacity.
- Space: O(nW) for the DP table.
